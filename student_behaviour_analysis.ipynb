```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Student_Behaviour.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
gender_count = df['Gender'].value_counts()

plt.bar(gender_count.index, gender_count.values)

plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Gender Distribution")

plt.show()
```

```python
df['Gender'].value_counts()
```

```python
plt.bar(gender_count.index, gender_count.values)
```

```python
plt.hist(df['Height(CM)'], bins=10)

plt.xlabel("Height")
plt.ylabel("Frequency")
plt.title("Height Distribution")

plt.show()
```

```python
plt.hist(df['Height(CM)'], bins=10)
```

```python
plt.hist(df['college mark'], bins=8)

plt.xlabel("College Marks")
plt.ylabel("Frequency")
plt.title("Distribution of College Marks")

plt.show()
```

```python
sns.boxplot(x='Gender', y='college mark', data=df)

plt.title("Gender vs College Marks")
plt.xlabel("Gender")
plt.ylabel("College Marks")

plt.show()
```

```python
sns.barplot(x='Department', y='college mark', data=df)

plt.title("Department vs College Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.show()
```

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])

sns.heatmap(numeric_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Certification Course',
    hue='Gender',
    data=df
)

plt.title("Status vs Gender Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Students")

plt.show()
```
